## Part 6 · Comparisons and decisions (if)
*About 25 min*

### 📘 LEARN · True, False and if

A comparison asks a yes/no question and gives **True** or **False** (called a *boolean*): `>` greater, `<` smaller, `>=` greater or equal,
`<=` smaller or equal, `==` equal (two equals signs; one `=` stores), `!=` not equal.
`if` lets Python choose what to do: the indented lines run only when the condition is True. `else` runs when it is not.

**📖 New words**

- **boolean**: a value that is True or False
- **condition**: a comparison that decides what happens
- **indentation**: spaces at the start of a line (4 spaces) that show which lines belong inside the if

### ▶️ SEE · Comparisons

Each line asks a question.

```python
sbp = 152
print(sbp >= 140)
print(sbp == 120)
print(sbp != 120)
```

**👀 What you should see:** `True`, `False`, `True`.

### ▶️ SEE · and / or

`and` needs both parts True; `or` needs at least one.

```python
age = 61
smoker = True
print(age >= 60 and smoker)
print(age >= 70 or smoker)
```

**👀 What you should see:** `True` and `True`.

### ▶️ SEE · if ... else

Practice cut-off: a reading of 140 or more is labelled 'High'. These labels are for practice only, not medical advice. Notice the colon at the end of the `if` and `else` lines and the four spaces before the indented lines.

```python
sbp = 152
if sbp >= 140:
    print("High reading")
else:
    print("Not high")
```

**🔎 What each line does**

- `if sbp >= 140:`  →  asks the question; the colon starts the block that runs when the answer is True
- `    print("High reading")`  →  indented, so it belongs to the `if`; runs only when `sbp >= 140` is True
- `else:`  →  what to do when the question was False
- `    print("Not high")`  →  indented, so it belongs to the `else`

**👀 What you should see:** `High reading`. Change `sbp` to 120 and run again to see the other branch.

### ▶️ SEE · Three or more choices with elif

`elif` (else if) adds more questions. Python checks them in order and stops at the first True.

```python
sbp = 130
if sbp >= 140:
    band = "High"
elif sbp >= 120:
    band = "Elevated"
else:
    band = "Normal"
print(band)
```

**🔎 What each line does**

- `if sbp >= 140:`  →  checked first
- `elif sbp >= 120:`  →  checked only if the first question was False
- `else:`  →  used when none of the questions was True
- `band = "Elevated"`  →  stores a label in a variable instead of printing it
- `print(band)`  →  shows the chosen label after the decisions are finished

**👀 What you should see:** `Elevated` (130 is below 140 but not below 120).

## Part 7 · Loops
*About 25 min*

### 📘 LEARN · Repeat work with a for loop

A **for loop** repeats the indented lines once for **each item** of a list. On every round, the loop variable (for example `age`) holds
the next item. Loops are how we process many values without copying code.

**📖 New words**

- **loop**: repeat steps
- **loop variable**: the name that holds the current item on each round
- **accumulator**: a variable that starts at 0 and grows during a loop

### ▶️ SEE · Visit every item

The loop visits each age once and prints it.

```python
ages = [54, 61, 47, 70]
for age in ages:
    print(age)
```

**🔎 What each line does**

- `for age in ages:`  →  for each item in the list `ages`, call it `age` and run the indented lines; do not forget the colon
- `    print(age)`  →  indented, so it runs once per item

**👀 What you should see:** four lines: 54, 61, 47, 70.

### ▶️ SEE · Add up with an accumulator

Start a total at 0, then add each age to it.

```python
ages = [54, 61, 47, 70]
total = 0
for age in ages:
    total = total + age
print(total)
```

**🔎 What each line does**

- `total = 0`  →  the accumulator starts at zero, before the loop
- `total = total + age`  →  on every round: old total plus the current age, stored back in `total`
- `print(total)`  →  not indented, so it runs once, after the loop has finished

**👀 What you should see:** `232`.

### ▶️ SEE · Count with a condition inside a loop

Combine the loop with an `if` to count only some items.

```python
ages = [54, 61, 47, 70]
n_older = 0
for age in ages:
    if age >= 60:
        n_older = n_older + 1
print(n_older)
```

**🔎 What each line does**

- `n_older = 0`  →  a counter that starts at zero
- `if age >= 60:`  →  asked on every round
- `n_older = n_older + 1`  →  indented twice: runs only when the `if` is True, adding 1 to the counter

**👀 What you should see:** `2` (the ages 61 and 70).

### ▶️ SEE · range: repeat a number of times

`range(3)` gives 0, 1, 2.

```python
for i in range(3):
    print(i)
```

**👀 What you should see:** `0`, `1`, `2` on separate lines.

### 📘 LEARN · Functions package a recipe

A **function** is a named recipe: you give it **inputs** (called *parameters*), it follows the steps, and it **returns** an answer.

**📖 New words**

### ▶️ SEE · Define and call a function

The first two lines *define* the recipe (nothing is calculated yet); the last line *calls* it.

```python
def bmi(weight_kg, height_m):
    return weight_kg / height_m ** 2

print(bmi(70, 1.75))
```

- `    return weight_kg / height_m ** 2`  →  indented: the body. It calculates the answer and hands it back with `return`
- `print(bmi(70, 1.75))`  →  calls the function with weight 70 and height 1.75, then prints what it returned

## Part 17 · Your first charts
*About 30 min · Research question(s): RQ2*

### 📘 LEARN · Why draw?

Numbers summarise; **charts show shape**. A **histogram** groups values into ranges (*bins*) and draws a bar for how many fall in each; it shows the
centre, the spread and any skew. A **bar chart** compares one number across categories. Every chart needs a **title** and **labelled axes**.
We use `plt`, the nickname for Matplotlib's plotting tools that was loaded in the setup.

**📖 New words**

- **histogram**: bars showing how many values fall in each range
- **bin**: one range of values in a histogram
- **axis label**: text saying what an axis shows (and its units)

### ▶️ SEE · A histogram of BMI

Missing values cannot be drawn, so `.dropna()` leaves them out **for the chart only**; the data itself is not changed.

```python
plt.figure(figsize=(6, 4))
plt.hist(df["BMI"].dropna(), bins=30)
plt.title("Body-mass index of the participants")
plt.xlabel("BMI (kg/m2)")
plt.ylabel("Number of people")
plt.show()
```

**🔎 What each line does**

- `plt.figure(figsize=(6, 4))`  →  starts a new blank chart 6 inches wide and 4 high
- `plt.hist(df["BMI"].dropna(), bins=30)`  →  draws the histogram of BMI with 30 bars, ignoring missing values
- `plt.title(`  →  adds the title
- `plt.xlabel(`  →  labels the horizontal axis with the variable and its unit
- `plt.ylabel(`  →  labels the vertical axis
- `plt.show()`  →  displays the finished chart

**👀 What you should see:** a bell-shaped histogram centred near 25-26 with a tail towards high BMI values.

### ▶️ SEE · A bar chart of CHD rate by smoking status

First we calculate the numbers, then we draw them. `.index` holds the group labels and `.values` the numbers.

```python
chd_rates = df.groupby("currentSmoker")["TenYearCHD"].mean()
plt.figure(figsize=(5, 4))
plt.bar(["Non-smoker", "Smoker"], chd_rates.values)
plt.title("Recorded CHD by smoking status")
plt.xlabel("Smoking status")
plt.ylabel("Proportion with CHD")
plt.show()
```

**🔎 What each line does**

- `chd_rates = df.groupby(`  →  the numbers to draw: the CHD rate for each smoking group
- `plt.bar(["Non-smoker", "Smoker"], chd_rates.values)`  →  one bar per group: the first list gives the bar names, the second the bar heights
- `plt.title(`  →  title
- `plt.xlabel(`  →  x label
- `plt.ylabel(`  →  y label
- `plt.show()`  →  display

**👀 What you should see:** two bars of similar height (0.145 and 0.159).
